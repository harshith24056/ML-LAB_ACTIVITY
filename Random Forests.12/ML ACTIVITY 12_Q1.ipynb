```python
from collections import Counter
import matplotlib.pyplot as plt

tree_votes = ["Apple", "Apple", "Banana"]  

vote_counts = Counter(tree_votes)
final_class = vote_counts.most_common(1)[0][0]

print("Votes per tree:", tree_votes)
print("Vote tally:", dict(vote_counts))
print("Final class after majority voting:", final_class) 
```

```python
plt.bar(vote_counts.keys(), vote_counts.values(), color=["#3498db", "#e67e22"])
plt.ylabel("Number of trees voting for this class")
plt.title("Random Forest Majority Voting")
plt.axhline(len(tree_votes) / 2, color="gray", linestyle="--", label="50% threshold")
plt.legend()
plt.show() 
```

```python
def majority_vote(votes):
    counts = Counter(votes)
    winner, n_votes = counts.most_common(1)[0]
    return winner, n_votes, len(votes)

larger_forest_votes = ["Apple", "Apple", "Banana", "Apple", "Banana", "Apple", "Apple"]
winner, n_votes, total = majority_vote(larger_forest_votes)
print(f"Final prediction: {winner} ({n_votes}/{total} trees agreed)")  
```
