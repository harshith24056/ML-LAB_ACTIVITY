```python
import pandas as pd
import numpy as np 
```

```python
data = pd.DataFrame({
    'Brightness': [40, 50, 60, 10, 70, 60, 25],
    'Saturation': [20, 50, 90, 25, 70, 10, 80],
    'Class': ['Red', 'Blue', 'Blue', 'Red', 'Blue', 'Red', 'Blue']
})

print(data)  
```

```python
new_point = {'Brightness': 45, 'Saturation': 45}

def euclidean_distance(x1, y1, x2, y2):
    return np.sqrt((x2 - x1)**2 + (y2 - y1)**2) 
```

```python
data['Distance'] = data.apply(
    lambda row: euclidean_distance(
        new_point['Brightness'], new_point['Saturation'],
        row['Brightness'], row['Saturation']
    ),
    axis=1
)
data_sorted = data.sort_values(by='Distance').reset_index(drop=True)
print(f"New point: Brightness={new_point['Brightness']}, Saturation={new_point['Saturation']}")
print("\nDistances (sorted, nearest first):")
print(data_sorted)  
```

```python
K = 3
nearest_k = data_sorted.head(K)
predicted_class = nearest_k['Class'].mode()[0]

print(f"K={K} nearest neighbors:")
print(nearest_k[['Brightness', 'Saturation', 'Class', 'Distance']])
print(f"\nPredicted class for new point: {predicted_class}") 
```
