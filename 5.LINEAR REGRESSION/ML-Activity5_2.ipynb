```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn import metrics
import matplotlib.pyplot as plt

# Creating the Dataset (from the given table)
data = pd.DataFrame({
    'Weight': [79, 69, 73, 95, 82, 55, 69, 71, 64, 69],
    'Height': [1.80, 1.68, 1.82, 1.70, 1.87, 1.55, 1.50, 1.78, 1.67, 1.64],
    'Age':    [35, 39, 25, 60, 27, 18, 89, 42, 16, 52],
    'Gender': ['Male','Male','Male','Male','Male',
               'Female','Female','Female','Female','Female']
})

print(data.shape)
data.head(10) 
```

```python
print(data.isnull().sum())
data['Gender'] = data['Gender'].map({'Male': 1, 'Female': 0})
data.head() 
```

```python
x = data[['Height', 'Age', 'Gender']].values
y = data['Weight'].values
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42)
model = LinearRegression()
model.fit(x_train, y_train)
y_pred = model.predict(x_test)
print('Predicted:', y_pred)
print('Actual   :', y_test) 
```

```python
mse = metrics.mean_squared_error(y_test, y_pred)
rmse = mse ** 0.5
r2 = metrics.r2_score(y_test, y_pred)
print('MSE  :', mse)
print('RMSE :', rmse)
print('R2 Score :', r2)
print('Intercept:', model.intercept_)
print('Coefficients (Height, Age, Gender):', model.coef_) 
```

```python
new_person = [[1.75, 30, 1]]
predicted_weight = model.predict(new_person)
print('Predicted Weight:', predicted_weight) 
```
