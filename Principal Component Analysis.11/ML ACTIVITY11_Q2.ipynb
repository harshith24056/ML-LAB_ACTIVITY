```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
loadings = pd.DataFrame({
    'PC1': [0.190, 0.544, 0.782, 0.365, 0.585, 0.394, 0.985, 0.520, 0.142],
    'PC2': [0.017, 0.020, -0.605, 0.294, 0.085, -0.273, 0.126, 0.402, 0.150],
    'PC3': [0.207, 0.204, 0.144, 0.585, 0.234, 0.027, -0.111, 0.519, 0.239]
}, index=['Climate', 'Housing', 'Health', 'Crime', 'Transportation',
          'Education', 'Arts', 'Recreation', 'Economy'])

loadings 
```

```python
threshold = 0.5   

for pc in loadings.columns:
    strong = loadings[pc][loadings[pc].abs() >= threshold]
    strong = strong.sort_values(key=abs, ascending=False)
    print(f"{pc} is mainly driven by:")
    print(strong.to_string())
    print()  
```

```python
for pc in loadings.columns:
    top = loadings[pc].abs().idxmax()
    print(f"{pc}: strongest variable = {top} ({loadings.loc[top, pc]})") 
```

```python
plt.figure(figsize=(6, 5))
plt.imshow(loadings, cmap='coolwarm', aspect='auto', vmin=-1, vmax=1)
plt.colorbar(label='Loading')
plt.xticks(range(3), loadings.columns)
plt.yticks(range(len(loadings.index)), loadings.index)

for i in range(loadings.shape[0]):
    for j in range(loadings.shape[1]):
        plt.text(j, i, f"{loadings.iloc[i, j]:.2f}", ha='center', va='center', fontsize=9)

plt.title('PCA Loadings')
plt.show() 
```
